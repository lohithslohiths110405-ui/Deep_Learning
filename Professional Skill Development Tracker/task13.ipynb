```python
import torch
import torch.nn as nn

# Sample Time-Series Data
X = torch.randn(2, 1, 20)   # Batch=2, Channels=1, Sequence Length=20

# Simple TCN Model
class TCN(nn.Module):

    def __init__(self):
        super().__init__()

        self.conv1 = nn.Conv1d(1, 8, kernel_size=3, dilation=1, padding=2)
        self.conv2 = nn.Conv1d(8, 8, kernel_size=3, dilation=2, padding=4)
        self.conv3 = nn.Conv1d(8, 1, kernel_size=3, dilation=4, padding=8)

        self.relu = nn.ReLU()

    def forward(self, x):

        x = self.relu(self.conv1(x))
        x = self.relu(self.conv2(x))
        x = self.conv3(x)

        # Keep output length same as input
        return x[:, :, :20]

# Create Model
model = TCN()

# Prediction
output = model(X)

# Loss
loss = nn.MSELoss()
target = torch.randn(2, 1, 20)

error = loss(output, target)

print("Input Shape :", X.shape)
print("Output Shape:", output.shape)
print("Loss :", error.item())
```
